```python
import sys
import warnings
from cochem_scribe.interfaces import web_pdf

# 1. The CSS Bleed Sabotage
canvas_html = '<canvas id="webgl-canvas" width="4000" height="4000"></canvas>'
canvas_width = 4000
parent_width = 1000

# 2. Rendering Audit & Canvas Verification
# We expect a CSSBleedWarning and the output to be wrapped in a responsive container
with warnings.catch_warnings(record=True) as w:
    warnings.simplefilter("always")
    
    output_html = web_pdf.render_preview(canvas_html, canvas_width, parent_width)
    
    assert len(w) == 1, f"Expected 1 warning, got {len(w)}"
    assert issubclass(w[-1].category, web_pdf.CSSBleedWarning), "Warning is not CSSBleedWarning"
    
    assert 'class="responsive-figure-container"' in output_html, "Missing responsive-figure-container class in output HTML"
    assert canvas_html in output_html, "Original canvas HTML missing from wrapped output"
    
    print("CSSBleedWarning successfully caught!")
    print("Output HTML successfully wrapped with responsive container!")
    print(output_html)
```
