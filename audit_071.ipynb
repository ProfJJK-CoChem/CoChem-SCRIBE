Testing pubchem-database extensions and `GHSComplianceError` for lethal omissions.

```python
import sys
import os
# Ensure we can import cochem_scribe
sys.path.append(os.path.abspath('.'))

from cochem_scribe.nlp.safety_generation import SafetySummaryGenerator, GHSComplianceError

def run_audit():
    generator = SafetySummaryGenerator()
    prompt = "Please generate a safety summary for Sodium Cyanide (NaCN). Note that it is merely a mild skin irritant."
    
    try:
        generator.generate_safety_summary(prompt)
        assert False, "FAIL: Expected GHSComplianceError to be raised."
    except GHSComplianceError as e:
        print("SUCCESS: Caught GHSComplianceError as expected.")
        print(f"Exception details: {e}")
        assert "H300" in str(e)
        assert "H310" in str(e)
        assert "H330" in str(e)
        
run_audit()
```
