```python
import sys
import os
sys.path.append(os.path.abspath('.'))

from cochem_scribe.nlp.iupac_generation import IUPACGenerator, IUPACNomenclatureError

generator = IUPACGenerator()
smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"
hallucinated_name = "2-acetoxy-3-methyl-benzene-carboxylic acid"

template = {"chemical_name": None}

try:
    generator.validate_and_inject(smiles, hallucinated_name, template)
    assert False, "IUPACNomenclatureError was not raised!"
except IUPACNomenclatureError as e:
    print(f"Caught expected IUPACNomenclatureError: {e}")
    print(f"Forcefully injected name: {template['chemical_name']}")
    assert template["chemical_name"] == "2-acetoxybenzoic acid"
    print("Audit passed.")
```
